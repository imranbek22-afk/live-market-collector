```python
import requests
import pandas as pd
from io import StringIO

url = "https://stooq.com/q/d/l/?s=aapl.us&i=d"
resp = requests.get(url, timeout=30, headers={"User-Agent": "live-market-collector-notebook/0.1"})
print("status", resp.status_code, "bytes", len(resp.content))
raw = pd.read_csv(StringIO(resp.text))
raw.tail()
```
